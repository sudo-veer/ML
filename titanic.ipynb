```python
#Dataset hint: make Pclass, Sex, Age and Fare columns, and put some missing values in Age. Build Survived (0/1) from a score where females and a lower class number (1st class) increase survival, plus noise.

# Logistic Regression: predicts Survived (0/1). Missing Age is filled with median, Sex converted to numbers.
import numpy as np, pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score, confusion_matrix

np.random.seed(1)
n = 200
df = pd.DataFrame({"Pclass": np.random.randint(1, 4, n),
                   "Sex": np.random.choice(["male", "female"], n),
                   "Age": np.random.randint(1, 70, n).astype(float),
                   "Fare": np.random.uniform(5, 200, n)})
df.loc[np.random.choice(n, 20, replace=False), "Age"] = np.nan
score = (df.Sex == "female")*2 - df.Pclass*0.7 + np.random.normal(0, 1, n)
df["Survived"] = (score > -0.3).astype(int)

df["Age"] = df["Age"].fillna(df["Age"].median())
df["Sex"] = df["Sex"].map({"male": 0, "female": 1})

X = df.drop("Survived", axis=1)
y = df["Survived"]
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2, random_state=1)

model = LogisticRegression(max_iter=1000)
model.fit(X_train, y_train)
pred = model.predict(X_test)

print("Accuracy :", accuracy_score(y_test, pred))
print("Precision:", precision_score(y_test, pred))
print("Recall   :", recall_score(y_test, pred))
print("F1-Score :", f1_score(y_test, pred))
print(confusion_matrix(y_test, pred))
```
